## 0. 준비 / Setup

설치할 것은 없습니다. 아래에 쓰는 NumPy · pandas · Matplotlib · statsmodels 는 전부 Colab 에 이미 들어 있고, 자료도 노트북 안에 있어 내려받을 것이 없습니다. 무작위 숫자는 쓰지 않으므로 같은 선택이면 언제나 같은 결과가 나옵니다.
*Nothing to install. NumPy, pandas, Matplotlib and statsmodels all ship with Colab, and the data are inside the notebook. Nothing is random, so the same choices always give the same results.*

**위에서 아래로 차례로 실행하세요.** 런타임을 다시 시작했다면 **런타임 > 모두 실행** 으로 처음부터 돌리세요.

```python
#@title ▶ 준비 / Setup. 먼저 실행하세요 / Run this first { display-mode: "form" }
import warnings, json, hashlib, platform, subprocess, datetime
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import statsmodels
from statsmodels.tsa.stattools import adfuller, kpss, acf, pacf
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tools.sm_exceptions import InterpolationWarning, ModelWarning

np.set_printoptions(precision=3, suppress=True)
plt.rcParams["figure.figsize"] = (11, 3.4)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3
GREEN, GOLD, RED = "#1E4A32", "#D9AE3A", "#B34F3F"          # 7주차와 같은 색 / the Week 7 colours

M = 12                                                      # 계절 주기. 월별 자료의 연 주기 / seasonal period, a yearly cycle in monthly data
H = 12                                                      # 앞으로 내다볼 달 수 / months to forecast
LB_LAG = 24                                                 # Ljung–Box 시차. 덱과 같은 2 × 12 / Ljung–Box lag, 2 × 12 as on the deck
ORIGINS = ["2014-09", "2015-09", "2016-09", "2017-09"]      # rolling origin 의 예측 원점 / forecast origins for the rolling origin
NEAR_UNIT = 1.05                                            # AR 근의 크기가 이보다 작으면 단위근 근처로 본다 / an AR root smaller than this counts as near a unit root
NOTEBOOK = "A2 v11 (2026-10-14)"


def test_pair(x, regression="c", kpss_lags=4):
    """ADF 와 KPSS 를 함께 돌려 (ADF 통계량, ADF p, KPSS 통계량, KPSS p) 를 돌려준다. 7주차 실습과 같다.
    Run ADF and KPSS together and return (ADF statistic, ADF p, KPSS statistic, KPSS p). Same as the Week 7 lab."""
    x = pd.Series(x).dropna().to_numpy(dtype=float)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", InterpolationWarning)   # 표 끝의 p 경고. 아래 fmt_p 가 처리 / the table-end warning, handled by fmt_p
        warnings.simplefilter("ignore", FutureWarning)          # 새 statsmodels 의 반환 형식 예고. 숫자는 같다 / a notice about a future return type, same numbers
        adf_stat, adf_p = adfuller(x, autolag="AIC", regression=regression)[:2]
        kpss_stat, kpss_p = kpss(x, regression=regression, nlags=kpss_lags)[:2]
    return adf_stat, adf_p, kpss_stat, kpss_p


def fmt_p(p, test="adf"):
    """p 값을 셋째 자리까지 적는다. KPSS 는 표 끝에서 잘린다. 예 / e.g. '= 0.052', '< 0.01', '> 0.10'.
    Write a p-value to three decimals. KPSS p-values are cut at the ends of a table."""
    if test == "kpss" and p <= 0.01:
        return "< 0.01"
    if test == "kpss" and p >= 0.10:
        return "> 0.10"
    if p < 0.001:
        return "< 0.001"
    return f"= {p:.3f}"


def decision_row(adf_p, kpss_p):
    """덱 「Read the Two Tests Together」 결정표의 줄(1 ~ 4). 7주차 실습의 TODO 1 과 같은 함수.
    The row (1 to 4) of the deck's decision table. The same function as TODO 1 of the Week 7 lab."""
    adf_drops = adf_p < 0.05            # 작은 p 면 「단위근」 을 버린다 / a small p drops "keeps every shock"
    kpss_drops = kpss_p < 0.05          # 작은 p 면 「정상」 을 버린다 / a small p drops "stationary"
    if adf_drops and not kpss_drops:
        return 2
    if (not adf_drops) and kpss_drops:
        return 1
    if adf_drops and kpss_drops:
        return 3
    return 4


ROW_TEXT = {
    1: ("둘 다 「정상 아님」 쪽입니다. 그림이 직선을 따라 오르면 덱 「Can the Tests Tell Series A From Series B?」 를 먼저 보세요. 아니면 한 번 더 차분하고 다시 검정하세요.",
        "Both point to \"not stationary\". If the plot rises along a straight line, see the deck slide \"Can the Tests Tell Series A From Series B?\" first. Otherwise, difference once and test again."),
    2: ("둘 다 「정상」 쪽입니다. 차분을 멈춥니다.",
        "Both point to \"stationary\". Stop differencing."),
    3: ("두 검정이 엇갈립니다. 그림을 보세요. 차분한다면 시차 1 막대가 −0.5 근처인지 확인하세요.",
        "They disagree. Look at the plot. If you difference, check the lag-1 bar for −0.5."),
    4: ("두 검정이 분명한 답을 주지 않습니다. 그림과 ACF 를 보세요. 많은 시차에서 상관이 아주 높게 남아 있으면 한 번 더 차분하고 다시 검정해 보세요.",
        "The tests do not give a clear answer. Look at the plot and the ACF. If the correlations stay very high across many lags, consider another difference and check again."),
}


def print_tests(label, x):
    """검정 두 개와 결정표의 줄을 한 번에 찍고 (ADF p, KPSS p, 줄) 을 돌려준다.
    Print both tests and the decision-table row, and return (ADF p, KPSS p, row)."""
    _, ap, _, kp = test_pair(x)
    row = decision_row(ap, kp)
    print(f"{label}:  ADF p {fmt_p(ap)}   KPSS p {fmt_p(kp, 'kpss')}   ->  결정표 {row}째 줄 / decision table row {row}")
    print(f"    {ROW_TEXT[row][0]}")
    print(f"    {ROW_TEXT[row][1]}")
    return ap, kp, row


def difference(x, D, d):
    """계절 차분 D 번(시차 12)을 한 뒤 보통 차분 d 번(시차 1)을 한다. 순서는 결과를 바꾸지 않는다(덱).
    Take D seasonal differences (lag 12), then d ordinary differences (lag 1). The order does not change the result (deck)."""
    z = pd.Series(x, dtype=float)
    for _ in range(D):
        z = z - z.shift(M)
    for _ in range(d):
        z = z - z.shift(1)
    return z.dropna()


def mae(e):
    """오차의 절댓값 평균 / mean absolute error."""
    return float(np.mean(np.abs(e)))


def rmse(e):
    """오차 제곱 평균의 제곱근 / root mean squared error."""
    return float(np.sqrt(np.mean(np.square(e))))


def model_name(order, seasonal_order):
    p, d, q = order
    P, D, Q, m = seasonal_order
    return f"ARIMA({p},{d},{q})({P},{D},{Q}){m}"


def fit_candidate(x, order, seasonal_order):
    """SARIMA 를 맞추고, 적합이 믿을 만했는지 세 가지로 점검해 함께 돌려준다.
    Fit a SARIMA model and report three checks on whether the fit can be trusted."""
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")
        res = SARIMAX(np.asarray(x, dtype=float), order=order, seasonal_order=seasonal_order).fit(disp=False)
    fit_warnings = [w for w in caught if issubclass(w.category, (ModelWarning, RuntimeWarning))]
    converged = res.mle_retvals.get("converged") if isinstance(res.mle_retvals, dict) else None
    ar_min = float(np.min(np.abs(res.arroots))) if len(res.arroots) else float("nan")
    reasons = []
    if converged is not True:
        reasons.append("not_converged")
    if fit_warnings:
        reasons.append("estimation_warning")
    if np.isfinite(ar_min) and ar_min < NEAR_UNIT:
        reasons.append("near_unit_root")
    return {"res": res, "order": order, "seasonal_order": seasonal_order, "name": model_name(order, seasonal_order),
            "converged": converged, "ar_min": ar_min, "reasons": reasons}


REASON_TEXT = {
    "not_converged": ("최적화가 수렴하지 않았습니다", "the optimisation did not converge"),
    "estimation_warning": ("추정 중에 경고가 났습니다", "the estimation raised a warning"),
    "near_unit_root": ("AR 부분의 근이 1 에 매우 가깝습니다(단위근 근처)", "a root of the AR part is very close to 1, near a unit root"),
}


def warn_box(fit, who):
    """적합이 믿을 만하지 않을 때만 안내 상자를 찍는다. Python 오류가 아니다.
    Print a notice only when the fit may not be reliable. This is not a Python error."""
    if not fit["reasons"]:
        return
    ko = ", ".join(REASON_TEXT[r][0] for r in fit["reasons"])
    en = "; ".join(REASON_TEXT[r][1] for r in fit["reasons"])
    line = "-" * 96
    print(line)
    print(f" 적합 경고 / Model fitting warning   [{who}: {fit['name']}]")
    print(" 이 모형은 믿을 만하게 적합되지 않았을 수 있습니다. 계열이 정상인지 확인하고 Q1 의 차분 선택을 다시 생각해 보세요.")
    print(" 과제는 계속할 수 있습니다. Python 오류가 아닙니다.")
    print(" This model may not have been fitted reliably. Check whether the series is stationary and consider revising")
    print(" your differencing choice. You can still continue the assignment. This is not a Python error.")
    print(f" 이유 / reason: {ko}  /  {en}")
    print(line)


def std_resid(res):
    """표준화 잔차. 처음 시작 구간은 뺀다. statsmodels 의 잔차 검정이 쓰는 것과 같은 잔차다.
    Standardised residuals with the start-up months removed, the same residuals statsmodels' own test uses."""
    cut = int(np.maximum(res.loglikelihood_burn, res.nobs_diffuse))
    return np.asarray(res.filter_results.standardized_forecasts_error[0, cut:])


def ljung_box_p(res, lag=LB_LAG):
    """시차 lag 까지의 Ljung–Box p. 모형이 쓴 계수 수만큼 자유도를 뺀다(df_adjust).
    The Ljung–Box p-value up to lag, with the degrees of freedom reduced by the number of fitted coefficients."""
    return float(res.test_serial_correlation("ljungbox", df_adjust=True, lags=[lag])[0, 1, 0])


def bars(ax, r, band, title):
    """막대 그림. 띠 밖은 빨강, 띠는 금색 점선. 7주차와 같은 모양.
    Bar plot: red outside the band, the band as gold dashes. Same look as Week 7."""
    lags = np.arange(1, len(r) + 1)
    ax.bar(lags, r, width=0.5, color=[RED if abs(v) > band else GREEN for v in r])
    ax.axhline(band, color=GOLD, ls="--")
    ax.axhline(-band, color=GOLD, ls="--")
    ax.set_title(title)
    ax.set_xlabel("lag (months)")


def outside(r, band):
    """띠 밖 막대를 {시차: 값} 으로 / bars outside the band as {lag: value}."""
    return {k + 1: round(float(v), 2) for k, v in enumerate(r) if abs(v) > band}


RESULTS = {}                                                # 절마다 계산에 쓴 선택과 결과 / the choices and results each section used


def record(section, **values):
    RESULTS[section] = values


print("준비 완료 / Setup ready")
print("numpy", np.__version__, "· pandas", pd.__version__, "· statsmodels", statsmodels.__version__)
```

```python
#@title ▶ 코드 칸 점검 / Checks for the code cells. 고치지 마세요 / Do not edit { display-mode: "form" }
# 코드 1 ~ 6 의 결과를 statsmodels 와 위 셀의 값에 맞춰 보는 함수들입니다. 여러분이 고칠 곳은 없습니다.
# Functions that check Code 1 to 6 against statsmodels and the cells above. There is nothing here for you to edit.
from scipy.signal import detrend as _detrend

CODE_CELLS = ["Code1", "Code2", "Code3", "Code4", "Code5", "Code6"]
BEFORE_TEST = CODE_CELLS[:5]                                # 마지막 12개월을 열기 전에 통과할 칸 / cells to pass before the last 12 months open
BLANK = ("아직 None 인 빈칸이 있습니다. None 을 코드로 바꾸세요.", "A blank (None) is still there. Replace None with your code.")


def code_ok(name):
    """그 코드 칸이 이번 실행에서 통과했나 / did that code cell pass in this run?"""
    return RESULTS.get(name, {}).get("ok") is True


def _call(fname, *args):
    """학생 함수를 부른다. Python 오류가 나도 멈추지 않고 그 내용을 돌려준다.
    Call a student's function. On a Python error, return the message instead of stopping."""
    fn = globals().get(fname)
    if not callable(fn):
        return None, (f"함수 {fname} 가 없습니다. 함수 이름을 바꾸지 마세요.", f"There is no function {fname}. Do not rename it.")
    try:
        return fn(*args), None
    except Exception as e:  # noqa: BLE001
        msg = f"{type(e).__name__}: {e}"
        return None, (f"Python 오류가 났습니다. {msg}", f"Python raised an error. {msg}")


def _pad(s, width):
    """한글처럼 두 칸을 차지하는 글자를 세어 왼쪽 정렬로 채운다 / left-align, counting wide characters such as Hangul as two columns."""
    import unicodedata
    w = sum(2 if unicodedata.east_asian_width(ch) in "WF" else 1 for ch in s)
    return s + " " * max(0, width - w)


def _num(x):
    try:
        return float(x)
    except (TypeError, ValueError):
        return float("nan")


def _close(a, b, tol):
    return abs(_num(a) - _num(b)) <= tol


def _report(name, passed, why, info=(), last=False, **extra):
    """코드 칸 결과를 찍고 RESULTS 에 남긴다 / print a code cell's result and record it."""
    num = name[-1]
    line = "-" * 96
    print(line)
    if passed:
        print(f" 코드 {num} / Code {num}   통과 / passed")
        for s in info:
            print("   " + s)
    else:
        print(f" 코드 {num} / Code {num}   아직 통과하지 않았습니다 / not passed yet")
        for ko, en in why:
            print("   " + ko)
            if en != ko:
                print("   " + en)
        if last:
            print(" 이 칸이 통과해야 영수증이 READY 가 됩니다. / The receipt reads READY only after this cell passes.")
        else:
            print(" 이 칸이 통과해야 마지막 12개월이 열리고 영수증이 READY 가 됩니다. 아래 셀은 그대로 돌아갑니다.")
            print(" The last 12 months open, and the receipt reads READY, only after this cell passes. The cells below still run.")
    print(line)
    now = {k: globals().get(k) for k in ("D", "d", "p")}     # Q1 의 칸은 p 를 정하기 전에 돈다 / the Q1 cells run before p is set
    record(name, **{k: v for k, v in now.items() if v is not None}, ok=bool(passed), **extra)


def _safe(check):
    """점검이 예상 못 한 값에 걸려도 실행이 멈추지 않게 한다 / keep a check from stopping the run on an unexpected value."""
    def run():
        try:
            check()
        except Exception as e:  # noqa: BLE001
            msg = f"{type(e).__name__}: {e}"
            try:
                _report("Code" + check.__name__[-1], False,
                        [(f"결과를 점검하지 못했습니다. 함수가 돌려주는 값의 모양을 확인하세요. {msg}",
                          f"The result could not be checked. Check the form of what your function returns. {msg}")],
                        last=check.__name__.endswith("6"))
            except Exception:  # noqa: BLE001
                print(f" 코드 {check.__name__[-1]} / Code {check.__name__[-1]}: 점검하지 못했습니다 / could not be checked ({msg})")
                RESULTS["Code" + check.__name__[-1]] = {"ok": False}
    return run



@_safe
def check_code1():
    why, info = [], []
    for name, x in (("원계열 / original", train), (f"여러분이 변환한 계열 / your transformed series ({label})", z)):
        xs = np.asarray(x, dtype=float)
        ref = acf(xs, nlags=36, fft=False)[1:]
        out, err = _call("acf_by_hand", xs.copy(), 36)
        if err:
            why.append(err)
            break
        if out is None:
            why.append(BLANK)
            break
        mine = np.asarray(out, dtype=float)
        if mine.shape != (36,):
            why.append(("시차 1 ~ 36 의 36개 값을 돌려줘야 합니다.", "The function must return 36 values, lags 1 to 36."))
            break
        if np.max(np.abs(mine - ref)) > 1e-6:
            adj = acf(xs, nlags=36, fft=False, adjusted=True)[1:]
            raw = np.array([(xs[h:] * xs[:len(xs) - h]).sum() / (xs * xs).sum() for h in range(1, 37)])
            k = int(np.argmax(np.abs(mine - ref)))
            if np.max(np.abs(mine - adj)) < 1e-6:
                hint = ("n − h 로 나눴습니다. 2주차 정의는 언제나 n 으로 나눕니다.", "You divided by n − h. The Week 2 definition always divides by n.")
            elif np.max(np.abs(mine - raw)) < 1e-6:
                hint = ("평균을 빼지 않았습니다. dev = x − x̄ 부터 만드세요.", "The mean was not subtracted. Start from dev = x − x̄.")
            else:
                hint = (f"시차 {k + 1} 에서 여러분 {mine[k]:.4f}, statsmodels {ref[k]:.4f}.", f"At lag {k + 1}: yours {mine[k]:.4f}, statsmodels {ref[k]:.4f}.")
            why.append((f"{name}: ACF 가 statsmodels 와 다릅니다. {hint[0]}", f"{name}: the ACF differs from statsmodels. {hint[1]}"))
            break
        info.append(f"{name}:  ρ̂(1) {mine[0]:.2f}   ρ̂(12) {mine[11]:.2f}   (statsmodels {ref[0]:.2f}, {ref[11]:.2f})")
    passed = not why and len(info) == 2
    if passed:
        info.append("위 셀들이 그린 ACF 막대가 바로 이 식입니다 / the ACF bars drawn above are exactly this formula")
    _report("Code1", passed, why, info)


@_safe
def check_code2():
    why, info, table = [], [], []
    x = train.to_numpy(dtype=float)
    out, err = _call("remove_line", x.copy())
    if err:
        why.append(err)
    elif out is None:
        why.append(BLANK)
    elif np.shape(out) != x.shape or np.max(np.abs(np.asarray(out, dtype=float) - _detrend(x, type="linear"))) > 1e-6:
        why.append(("직선을 뺀 값이 기준 계산과 다릅니다. np.polyfit(t, x, 1) 이 [기울기, 절편] 순서로 돌려주는지 확인하세요.",
                    "What is left differs from the reference. Check that np.polyfit(t, x, 1) returns [slope, intercept] in that order."))
    elif not code_ok("Code1"):
        why.append(("직선 빼기는 맞습니다. 비교표는 코드 1 의 acf_by_hand 를 씁니다. 코드 1 부터 통과시키세요.",
                    "The line removal is right. The comparison table uses acf_by_hand from Code 1. Make Code 1 pass first."))
    else:
        rows = [("원계열 / original", x),
                ("직선을 뺀 계열 / line removed (계열 A 의 방법 / Series A's cure)", np.asarray(out, dtype=float)),
                (f"여러분의 차분 / your differences (D = {D}, d = {d})", difference(train, D, d).to_numpy()),
                ("여기서 보통 차분 한 번 더 / one more ordinary difference", difference(train, D, d + 1).to_numpy())]
        for name, s in rows:
            r_ = acf_by_hand(s, 12)
            table.append((name, float(r_[0]), float(r_[11]), len(s)))
        info.append("변환별 표본 ACF, 시차 1 · 12 와 값 개수 (여러분의 코드 1) / sample ACF at lags 1 and 12, and the count, for each transformation (your Code 1)")
        info.append("  " + _pad("", 70) + f"{'lag 1':>8}{'lag 12':>9}{'n':>6}")
        for name, a1, a12, n_ in table:
            info.append("  " + _pad(name, 70) + f"{a1:>8.2f}{a12:>9.2f}{n_:>6}")
        info.append("직선을 뺀 뒤에도 시차 1 이 1 에 가깝게 남으면, 충격이 쌓이는 계열(5주차 계열 B)에 가깝다는 신호입니다. Q1(b) 에 쓸 수 있습니다.")
        info.append("If the lag-1 value stays close to 1 after removing the line, the series behaves more like one that accumulates shocks (Week 5, Series B). You can use this in Q1(b).")
        info.append("한 번 더 차분한 줄의 시차 1 이 −0.5 근처면, 그 차분은 하나 많았을 수 있습니다(7주차 「What If We Difference One Time Too Many?」).")
        info.append("If the extra difference gives a lag-1 value near −0.5, that difference may be one too many (Week 7, \"What If We Difference One Time Too Many?\").")
    passed = not why and len(table) == 4
    _report("Code2", passed, why, info, table=[(round(a1, 2), round(a12, 2)) for _, a1, a12, _ in table])


@_safe
def check_code3():
    why, info = [], []
    zs = np.asarray(z, dtype=float)
    r_ = acf(zs, nlags=2, fft=False)
    r1, r2 = float(r_[1]), float(r_[2])
    ref = pacf(zs, nlags=2, method="ywm")
    out, err = _call("pacf_by_hand", r1, r2)
    if err:
        why.append(err)
    elif not isinstance(out, (tuple, list)) or len(out) != 2 or any(v is None for v in out):
        why.append(BLANK)
    else:
        p1, p2 = _num(out[0]), _num(out[1])
        if not _close(p1, ref[1], 1e-8):
            why.append((f"PACF(1) 은 ρ̂(1) 그대로입니다(여러분 {p1:.4f}, 맞는 값 {ref[1]:.4f}).", f"PACF(1) is just ρ̂(1) (yours {p1:.4f}, correct {ref[1]:.4f})."))
        elif not _close(p2, ref[2], 1e-8):
            hint = (("ACF(2) 를 그대로 썼습니다. 지난달을 거쳐 오는 몫 ρ̂(1)² 을 빼세요.", "That is ACF(2) itself. Remove the part carried through last month, ρ̂(1)².") if _close(p2, r2, 1e-8) else
                    ("1 − ρ̂(1)² 로 나누는 것이 빠졌습니다.", "The division by 1 − ρ̂(1)² is missing.") if _close(p2, r2 - r1 ** 2, 1e-8) else
                    ("ρ̂(1) 을 제곱하세요.", "Square ρ̂(1).") if _close(p2, (r2 - r1) / (1 - r1), 1e-8) else ("", ""))
            why.append((f"PACF(2) 가 위 셀의 PACF 와 다릅니다(여러분 {p2:.4f}, 맞는 값 {ref[2]:.4f}). {hint[0]}",
                        f"PACF(2) differs from the PACF in the cell above (yours {p2:.4f}, correct {ref[2]:.4f}). {hint[1]}"))
        else:
            info.append(f"여러분이 변환한 계열 / your transformed series ({label}):   ρ̂(1) {r1:.2f}   ρ̂(2) {r2:.2f}")
            info.append(f"PACF(1) {p1:.2f}   PACF(2) {p2:.2f}   (위 셀 / the cell above {ref[1]:.2f}, {ref[2]:.2f})")
            info.append(f"ACF(2) {r2:.2f} 가운데 지난달을 거쳐 오는 몫은 ρ̂(1)² = {r1 ** 2:.2f} 입니다. PACF(2) 는 그것을 빼고 남은 몫입니다.")
            info.append(f"Of ACF(2) = {r2:.2f}, the part carried through last month is ρ̂(1)² = {r1 ** 2:.2f}. PACF(2) is what is left after removing it.")
    _report("Code3", not why and bool(info), why, info, pacf1=round(float(ref[1]), 2), pacf2=round(float(ref[2]), 2))


@_safe
def check_code4():
    why, info, table = [], [], []
    rows, err = _call("candidate_table")
    if err:
        why.append(err)
    elif not isinstance(rows, list) or len(rows) != 4:
        why.append(("candidate_table 은 줄 네 개(p = 0 ~ 3)를 돌려줘야 합니다.", "candidate_table must return four rows (p = 0 to 3)."))
    elif any(r.get("fit") is None or r.get("values") is None for r in rows):
        why.append(BLANK)
    else:
        for r in rows:
            pp, fit, vals = r["p"], r["fit"], r["values"]
            res = fit.get("res") if isinstance(fit, dict) else None
            if res is None:
                why.append(("fit 에는 fit_candidate(...) 가 돌려준 것을 그대로 넣으세요.", "Put exactly what fit_candidate(...) returns into fit."))
                break
            if res.nobs == len(y):
                why.append((f"p = {pp}: 전체 {len(y)}개월(y)로 맞췄습니다. 마지막 12개월이 들어가면 안 됩니다. train 으로 맞추세요.",
                            f"p = {pp}: fitted to all {len(y)} months (y). The last 12 months must stay out. Fit to train."))
                break
            if res.nobs != len(train):
                why.append((f"p = {pp}: 훈련 {len(train)}개월로 맞추세요(지금 {res.nobs}개월).",
                            f"p = {pp}: fit to the {len(train)} training months (now {res.nobs})."))
                break
            if tuple(fit["order"]) != (pp, d, 0) or tuple(fit["seasonal_order"]) != tuple(SEAS):
                why.append((f"p = {pp}: 차수가 다릅니다({fit['name']}). order 와 SEAS 를 그대로 쓰세요.",
                            f"p = {pp}: the orders differ ({fit['name']}). Use order and SEAS as they are."))
                break
            if not isinstance(vals, (tuple, list)) or len(vals) != 2 or not _close(vals[0], res.aic, 0.01) or not _close(vals[1], res.bic, 0.01):
                why.append((f"p = {pp}: values 에는 이 적합의 (AIC, BIC) 를 그 순서로 넣으세요.", f"p = {pp}: values must hold this fit's (AIC, BIC), in that order."))
                break
            table.append((pp, fit["name"], _num(vals[0]), _num(vals[1]), "OK" if not fit["reasons"] else "경고 / warning"))
    passed = not why and len(table) == 4
    if passed:
        info.append(f"후보 비교표 / candidate table   (훈련 {len(train)}개월 / {len(train)} training months)")
        info.append(f"  {'p':>2}  " + _pad("모형 / model", 24) + f"{'AIC':>10}{'BIC':>10}   적합 / fit")
        for pp, name, aic, bic, note in table:
            mark = "   <- 여러분의 B / your B" if pp == p else ""
            info.append(f"  {pp:>2}  {name:<24}{aic:>10.2f}{bic:>10.2f}   {note}{mark}")
        info.append(f"  후보 A / candidate A   {FIT_A['name']}   AIC {FIT_A['res'].aic:.2f}   BIC {FIT_A['res'].bic:.2f}")
        info.append("AIC · BIC 가 가장 낮은 p 를 그대로 고르라는 표가 아닙니다. Q2(b) 의 p 는 PACF 로 정하고, 이 표로 이웃 차수와 견주어 점검합니다.")
        info.append("This table is not a rule to take the lowest AIC or BIC. Choose p in Q2(b) from the PACF, and use the table to check it against the neighbouring orders.")
    _report("Code4", passed, why, info, table=[(t[0], round(t[2], 2), round(t[3], 2)) for t in table])


@_safe
def check_code5():
    why, info, cover = [], [], []
    for i, org in enumerate(ORIGINS):
        row = [org]
        for who, order in (("A", (0, d, 3)), ("B", (p, d, 0))):
            out, err = _call("evaluate_at", train, org, order)
            if err:
                why.append(err)
                break
            if out is None:
                why.append(BLANK)
                break
            if not isinstance(out, dict) or not {"past_end", "rmse", "inside"} <= set(out):
                why.append(("evaluate_at 의 마지막 return 줄은 고치지 마세요.", "Do not edit the last return line of evaluate_at."))
                break
            if out["past_end"] != org:
                why.append((f"원점 {org}: past 가 {out['past_end']} 에서 끝납니다. 원점 {org} 에서 끝나야 합니다. 원점 뒤 자료로 맞추면 미래를 보고 맞춘 셈입니다(누설).",
                            f"origin {org}: past ends at {out['past_end']}; it must end at the origin {org}. Fitting with data after the origin means using the future (leakage)."))
                break
            fut = train.loc[str(pd.Period(org, "M") + 1):].iloc[:H].to_numpy()
            fc_ = fit_candidate(train.loc[:org], order, SEAS)["res"].get_forecast(H)
            e_ = fut - np.asarray(fc_.predicted_mean)
            ref_rmse = roll_rows[i][who]
            if not _close(out["rmse"], ref_rmse, 0.01):
                v = _num(out["rmse"])
                hint = (("제곱근이 빠졌습니다. 그것은 MSE 입니다.", "The square root is missing; that is the MSE.") if _close(v, np.mean(e_ ** 2), 0.01) else
                        ("그것은 MAE 입니다. 오차를 제곱해 평균한 뒤 제곱근을 씌웁니다.", "That is the MAE. Square the errors, average them, then take the square root.") if _close(v, np.mean(np.abs(e_)), 0.01) else
                        ("fc 는 past 로 맞춘 모형의 예측이어야 하고, err 는 실제값 − 예측값입니다.", "fc must come from the model fitted to past, and err is actual − forecast."))
                why.append((f"원점 {org}, 후보 {who}: RMSE 가 위 rolling 표와 다릅니다(여러분 {v:.2f}, 표 {ref_rmse:.2f}). {hint[0]}",
                            f"origin {org}, candidate {who}: the RMSE differs from the rolling table above (yours {v:.2f}, table {ref_rmse:.2f}). {hint[1]}"))
                break
            ci = np.asarray(fc_.conf_int(alpha=0.05))
            ref_in = int(((fut >= ci[:, 0]) & (fut <= ci[:, 1])).sum())
            if out["inside"] != ref_in:
                why.append((f"원점 {org}, 후보 {who}: 구간 안 개수가 다릅니다(여러분 {out['inside']}, 맞는 값 {ref_in}). inside 줄을 확인하세요.",
                            f"origin {org}, candidate {who}: the count inside differs (yours {out['inside']}, correct {ref_in}). Check the inside line."))
                break
            row.append(ref_in)
        if why:
            break
        cover.append(tuple(row))
    passed = not why and len(cover) == len(ORIGINS)
    total = H * len(ORIGINS)
    t_a, t_b = sum(r[1] for r in cover), sum(r[2] for r in cover)
    if passed:
        info.append("원점마다 RMSE 가 위 rolling 표와 같습니다 / the RMSE matches the rolling table at every origin")
        info.append(f"rolling 95% 구간 안 / inside the 95% interval   (원점마다 {H}개월 / {H} months per origin)")
        info.append("  원점 / origin        A          B")
        for org, a_, b_ in cover:
            info.append(f"  {org}            {a_:>2} / {H}    {b_:>2} / {H}")
        info.append(f"  합 / total         {t_a:>2} / {total}    {t_b:>2} / {total}")
        info.append(f"95% 구간이 제대로 맞는다면 {total}개 가운데 평균 {0.95 * total:.1f}개쯤이 안에 듭니다. 같은 원점의 {H}개월은 서로 이어져 있어 독립된 {total}번의 점검은 아닙니다.")
        info.append(f"If the 95% intervals are right, about {0.95 * total:.1f} of {total} forecasts fall inside on average. The {H} months from one origin are linked, so these are not {total} independent checks.")
        info.append("Q4(c) · (d) 의 근거로 쓸 수 있습니다 / You can use this in Q4(c) and (d).")
    _report("Code5", passed, why, info, cover=cover, total_A=t_a, total_B=t_b, total=total)


@_safe
def check_code6():
    if preferred not in ("A", "B") or not all(code_ok(c) for c in BEFORE_TEST):
        line = "-" * 96
        print(line)
        print(" 코드 6 / Code 6   마지막 12개월이 열린 뒤에 점검합니다 / checked after the last 12 months open")
        print(line)
        record("Code6", D=D, d=d, p=p, ok=False, pending=True)
        return
    why, info, ends = [], [], {}
    chosen = FIT_A if preferred == "A" else FIT_B
    out, err = _call("report_table", chosen, test.copy())
    fc = chosen["res"].get_forecast(H)
    mean = np.asarray(fc.predicted_mean)
    ci = np.asarray(fc.conf_int(alpha=0.05))
    actual = test.to_numpy()
    ins = (actual >= ci[:, 0]) & (actual <= ci[:, 1])
    cols = ["forecast", "lower", "upper", "width", "actual", "inside"]
    if err:
        why.append(err)
    elif out is None:
        why.append(BLANK)
    elif not isinstance(out, pd.DataFrame):
        why.append(("report 는 pd.DataFrame 으로 만든 표여야 합니다.", "report must be a table made with pd.DataFrame."))
    elif list(out.columns) != cols:
        why.append(("열 이름과 순서를 forecast, lower, upper, width, actual, inside 로 하세요.", "Name and order the columns forecast, lower, upper, width, actual, inside."))
    elif len(out) != H or list(out.index) != list(test.index):
        why.append(("한 줄에 한 달씩 12줄이고, 행 이름은 test.index 여야 합니다.", "One row per month, 12 rows, with test.index as the row labels."))
    elif not (np.allclose(out["forecast"].to_numpy(float), mean) and np.allclose(out["lower"].to_numpy(float), ci[:, 0])
              and np.allclose(out["upper"].to_numpy(float), ci[:, 1]) and np.allclose(out["actual"].to_numpy(float), actual)):
        why.append(("값이 위 2단계 그림의 예측값 · 95% 구간 · 실제값과 다릅니다. 열마다 무엇을 넣었는지 확인하세요.",
                    "The values differ from the forecasts, 95% intervals or actual values in the step 2 plot. Check what went into each column."))
    elif not np.allclose(out["width"].to_numpy(float), ci[:, 1] - ci[:, 0]):
        why.append(("width 는 위 끝 − 아래 끝입니다.", "width is the upper end minus the lower end."))
    elif list(out["inside"].astype(bool)) != list(ins):
        why.append(("inside 열은 그 달 실제값이 구간 안이면 True 입니다.", "The inside column is True when that month's actual value is inside the interval."))
    else:
        show = out.copy()
        show.index = [str(t) for t in show.index]
        info.append(f"보고표 / report table   {chosen['name']}  (잠정 선택 / tentative choice {preferred}, 천 명 / thousands)")
        info.extend(show.round(1).to_string().split("\n"))
        w = ci[:, 1] - ci[:, 0]
        info.append(f"구간 폭이 h = 1 의 {w[0]:.1f} 에서 h = 12 의 {w[-1]:.1f} 로 넓어집니다(7주차 「Why Do the Intervals Keep Widening?」).")
        info.append(f"The interval widens from {w[0]:.1f} at h = 1 to {w[-1]:.1f} at h = 12 (Week 7, \"Why Do the Intervals Keep Widening?\").")
        info.append(f"구간 안 / inside: {int(ins.sum())} / {H}. 인력 계획에는 예측값 한 줄보다 이 아래 · 위 끝이 범위로 쓰입니다(Q4(d)).")
        info.append(f"Inside: {int(ins.sum())} of {H}. For workforce planning, the lower and upper ends give a range, not just one number (Q4(d)).")
        ends = {"first_m": (str(test.index[0]), float(ci[0, 0]), float(ci[0, 1])), "last_m": (str(test.index[-1]), float(ci[-1, 0]), float(ci[-1, 1]))}
    _report("Code6", not why and bool(ends), why, info, last=True, preferred=preferred, **ends)


print("코드 칸 점검 준비 완료 / Code-cell checks ready")
```

Ljung–Box 는 시차 1 ~ 24 의 자기상관을 **한꺼번에** 검정합니다. 출발 가정은 「남은 자기상관이 없다」 이고, 큰 p 는 「뚜렷한 무늬를 찾지 못했다」 는 뜻입니다. 띠 ±2/√n 은 **시차마다 따로** 긋는 기준입니다. 잔차는 표준화 잔차(그 달의 예측 오차를 예측 표준편차로 나눈 값)이고, 모형이 시작값에서 출발하는 처음 몇 달은 뺍니다. Ljung–Box 와 그림은 같은 잔차로 계산합니다.
*Ljung–Box tests the autocorrelations at lags 1 to 24 **together**. Its starting assumption is "no autocorrelation is left", and a large p means "no clear pattern was found". The band ±2/√n is drawn **for each lag separately**. The residuals are standardised (each month's forecast error divided by its forecast standard deviation), with the start-up months removed. The test and the plot use the same residuals.*

```python
#@title Q3 ▶ 두 후보의 잔차 / Residuals of both candidates { display-mode: "form" }
fig, ax = plt.subplots(1, 2, figsize=(12, 3.4), sharey=True)
q3 = {}
for a_, fit, who in [(ax[0], FIT_A, "A"), (ax[1], FIT_B, "B")]:
    e = std_resid(fit["res"])
    r = acf(e, nlags=LB_LAG, fft=False)[1:]
    band = 2 / np.sqrt(len(e))
    bars(a_, r, band, f"residual ACF, candidate {who}: {fit['name']}")
    q3[who] = {"lb": ljung_box_p(fit["res"]), "out": sorted(outside(r, band)), "band": band, "n": len(e)}
plt.tight_layout()
plt.show()

for who, fit in [("A", FIT_A), ("B", FIT_B)]:
    print(f"후보 {who} / candidate {who}  {fit['name']}:  Ljung–Box p (시차 / lag {LB_LAG}) {fmt_p(q3[who]['lb'])}"
          f"   띠 밖 시차 / lags outside the band {q3[who]['out']}   (잔차 / residuals {q3[who]['n']}, 띠 / band ±{q3[who]['band']:.3f})")
    warn_box(fit, f"후보 {who} / candidate {who}")
record("Q3", D=D, d=d, p=p, lb_A=q3["A"]["lb"], lb_B=q3["B"]["lb"], out_A=q3["A"]["out"], out_B=q3["B"]["out"])
```

```python
#@title Q4 ▶ 1단계. rolling origin / Step 1. Rolling origin { display-mode: "form" }
roll_rows, warned = [], 0
for org in ORIGINS:
    past = train.loc[:org]                                   # 원점까지의 자료만 / only the data up to the origin
    future = train.loc[str(pd.Period(org, "M") + 1):].iloc[:H]
    fa = fit_candidate(past, (0, d, 3), SEAS)
    fb = fit_candidate(past, (p, d, 0), SEAS)
    warned += bool(fa["reasons"]) + bool(fb["reasons"])
    roll_rows.append({
        "원점 / origin": org,
        "시험 / tested": f"{future.index[0]} ~ {future.index[-1]}",
        "A": rmse(future.to_numpy() - fa["res"].get_forecast(H).predicted_mean),
        "B": rmse(future.to_numpy() - fb["res"].get_forecast(H).predicted_mean),
        "seasonal naive": rmse(future.to_numpy() - past.iloc[-M:].to_numpy()),
    })
roll = pd.DataFrame(roll_rows)
mean_row = {"원점 / origin": "평균 / mean", "시험 / tested": "",
            **{k: roll[k].mean() for k in ["A", "B", "seasonal naive"]}}
roll = pd.concat([roll, pd.DataFrame([mean_row])], ignore_index=True)

print(f"RMSE (천 명 / thousands).  A = {FIT_A['name']}   B = {FIT_B['name']}")
print(roll.round(2).to_string(index=False))
if warned:
    print(f"\n적합 경고가 난 적합 / fits with a fitting warning : {warned} / {2 * len(ORIGINS)}. 이 숫자는 조심해서 읽으세요. / Read these numbers with care.")
record("Q4_roll", D=D, d=d, p=p, mean_A=float(roll["A"].iloc[-1]), mean_B=float(roll["B"].iloc[-1]),
       mean_sn=float(roll["seasonal naive"].iloc[-1]), warned_fits=int(warned),
       by_origin=[(r_["원점 / origin"], round(r_["A"], 2), round(r_["B"], 2), round(r_["seasonal naive"], 2)) for r_ in roll_rows])
```

### 코드 5 / Code 5. 원점 하나 평가하기 / Evaluating one forecast origin

With your function the notebook collects 4 origins × 12 months = 48 forecasts. The last 12 months are still not used.*

$$\mathrm{RMSE} = \sqrt{\frac{1}{m}\sum_{i=1}^{m} e_i^{2}}, \qquad e_i = y_i - \hat y_i$$

$y_i$ 는 실제값, $\hat y_i$ 는 예측값, $m$ 은 예측 개수(12)입니다.
*$y_i$ is the actual value, $\hat y_i$ the forecast and $m$ the number of forecasts (12).*
